```python
import pandas as pd

# Read data
text_df = pd.read_csv('Google_data (2b.c1).csv')
excel_df = pd.read_excel('data (2c2).xlsx', sheet_name='Sheet1')
web_df = pd.read_csv('https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv') # Replace with actual URL

# Display data
print("Text DataFrame Head:\n", text_df.head(), "\n")
print("Excel DataFrame Head:\n", excel_df.head(), "\n")
print("Web DataFrame Head:\n", web_df.head(), "\n")

# Handle missing values using modern methods
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# Save processed data
text_df.to_csv('processed_text.csv', index=False)
excel_df.to_excel('processed_excel.xlsx', index=False)

print("Processed data saved to 'processed_text.csv' and 'processed_excel.xlsx'.")
```
