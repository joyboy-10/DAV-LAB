```python
import pandas as pd
import numpy as np

# Load dataset into a DataFrame
df = pd.read_csv('data.csv')

# --- Start of fix for TypeError: Could not convert to numeric ---
# Convert 'Reviews' to numeric
df['Reviews'] = pd.to_numeric(df['Reviews'])

# Convert 'Size' to numeric (handling 'M', 'k', and 'Varies with device')
def clean_size(size_str):
    if isinstance(size_str, str):
        if 'M' in size_str:
            return float(size_str.replace('M', '')) * 10**6
        elif 'k' in size_str:
            return float(size_str.replace('k', '')) * 10**3
        elif 'Varies with device' in size_str:
            return np.nan # Treat 'Varies with device' as NaN
    return float(size_str)

df['Size'] = df['Size'].apply(clean_size)

# Convert 'Installs' to numeric
df['Installs'] = df['Installs'].str.replace('+', '', regex=False).str.replace(',', '', regex=False).astype(int)

# Convert 'Price' to numeric (assuming it's '0' or a numerical string)
df['Price'] = pd.to_numeric(df['Price'])
# --- End of fix ---

# Display first and last few rows
print("First 5 rows:\n", df.head())
print("Last 5 rows:\n", df.tail())
# Check data types and general info
df.info()
# Summary statistics
print("Summary statistics:\n", df.describe())
# Handle missing values (now df.mean() will work on numeric columns)
df.fillna(df.mean(numeric_only=True), inplace=True)

# --- Placeholder column operations (now fixed with actual columns) ---

# Create a new column based on 'Rating'
df['rating_doubled'] = df['Rating'] * 2

# Create a Series and perform operations using 'Rating'
series = df['Rating']
print("Series addition (Rating + 10):\n", series + 10)

# Filter rows based on 'Reviews' and 'Installs'
# Filtering for Apps with more than 50 reviews and less than 100,000 installs
filtered_df = df[(df['Reviews'] > 50) & (df['Installs'] < 100000)]
print("Filtered DataFrame (Reviews > 50 & Installs < 100000):\n", filtered_df)

# Grouping and aggregation by 'Category' to find mean 'Rating'
grouped = df.groupby('Category')['Rating'].mean()
print("Grouped mean Rating by Category:\n", grouped)

# Sorting by 'Rating' in descending order
df_sorted = df.sort_values(by='Rating', ascending=False)
print("Sorted DataFrame by Rating (descending):\n", df_sorted)

# Boolean masking for Apps with 'Reviews' greater than the median 'Reviews'
masked_df = df[df['Reviews'] > df['Reviews'].median()]
print("Masked DataFrame (Reviews > median Reviews):\n", masked_df)

# Remove duplicates and drop missing values
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# Create a new DataFrame with selected columns 'App' and 'Category'
subset_df = df[['App', 'Category']]
print("Subset DataFrame (App, Category):\n", subset_df.head())

# Save the new DataFrame to a CSV file
subset_df.to_csv('filtered_data.csv', index=False)

# Compute summary statistics for 'Installs'
print("Total sum of Installs:", df['Installs'].sum())
print("Mean of Installs:", df['Installs'].mean())
print("Standard Deviation of Installs:", df['Installs'].std())
```
